## Verify the OpenAI API key

```python
import os
assert os.getenv("OPENAI_API_KEY"), "Set OPENAI_API_KEY in your environment"
```

## Configure the model

We use OpenAI `gpt-5-mini`, with the API key read from the `OPENAI_API_KEY` environment variable (set `OPENAI_MODEL_ID` to use another model).

```python
import os
MODEL_ID = os.getenv("OPENAI_MODEL_ID", "gpt-5-mini")
os.environ["BYPASS_TOOL_CONSENT"] = os.getenv("BYPASS_TOOL_CONSENT", "true")
os.environ["STRANDS_TOOL_CONSOLE_MODE"] = "enabled"
print(f"Model:  {MODEL_ID}")
```

## Install Strands (first time only)

```python
%pip install -q 'strands-agents[openai]' strands-agents-tools
```

## The agent

```python
from strands import Agent
from strands.models.openai import OpenAIModel
from strands_tools import shell
model = OpenAIModel(
    client_args={"api_key": os.environ["OPENAI_API_KEY"]},
    model_id=MODEL_ID,
)
agent = Agent(
    model=model,
    tools=[shell],
    system_prompt="You are a helpful assistant. Use the shell tool when useful. Be concise."
)
```

## Ask it things

```python
agent("hello, introduce yourself")
```

```python
agent("what files are in this directory?")
```

### What just happened?The model decided to call `shell(ls)` on its own. That's Strands' model-first philosophy.
